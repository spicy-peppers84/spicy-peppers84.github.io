```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 100)
y = np.sqrt(x**2) ** 3

fig, ax = plt.subplots()
ax.plot (x, y)
ax.set_xlabel("x")
ax.set_ylabel("y")

ax.set_xbound(-10, 10)
ax.set_ybound(-100, 100)

plt.grid(True)
plt.show()
```
